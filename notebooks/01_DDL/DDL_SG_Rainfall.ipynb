# DDL Salto Grande - Precipitacion

Crea estructuras Raw, Bronze y Silver para la fuente SG sin mezclarla con otras fuentes.

```python
spark.sql('CREATE CATALOG IF NOT EXISTS weather')
spark.sql('CREATE SCHEMA IF NOT EXISTS weather.raw')
spark.sql('CREATE SCHEMA IF NOT EXISTS weather.bronze')
spark.sql('CREATE SCHEMA IF NOT EXISTS weather.silver')
spark.sql('CREATE VOLUME IF NOT EXISTS weather.raw.sg_volume')

dbutils.fs.mkdirs('/Volumes/weather/raw/sg_volume/json/daily')
```

```python
spark.sql('''
CREATE TABLE IF NOT EXISTS weather.bronze.sg_rainfall (
  fecha DATE,
  fecha_raw STRING,
  id_estacion STRING,
  nombre STRING,
  latitud DOUBLE,
  longitud DOUBLE,
  p DOUBLE,
  source_api STRING,
  source_file STRING,
  extracted_at TIMESTAMP,
  ingestion_date DATE,
  loaded_at TIMESTAMP,
  updated_at TIMESTAMP
)
USING DELTA
''')

spark.sql('''
CREATE TABLE IF NOT EXISTS weather.silver.sg_rainfall_daily (
  fecha DATE,
  id_estacion STRING,
  nombre STRING,
  latitud DOUBLE,
  longitud DOUBLE,
  lluvia_acumulada_mm DOUBLE,
  registros_total BIGINT,
  registros_validos BIGINT,
  source_table STRING,
  processed_at TIMESTAMP,
  updated_at TIMESTAMP
)
USING DELTA
''')
```

```python
for table_name in ['weather.bronze.sg_rainfall', 'weather.silver.sg_rainfall_daily']:
    print(f'DESCRIBE {table_name}')
    spark.sql(f'DESCRIBE {table_name}').show(truncate=False)
```
